```python
# Create a 1D NumPy array
array_1d = np.array([10, 20, 30, 40, 50])

# Create a 2D NumPy array
array_2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# Create a 3D NumPy array
array_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

# Display arrays
print("1D Array:")
print(array_1d)

print("\n2D Array:")
print(array_2d)

print("\n3D Array:")
print(array_3d)

# Verify shapes
print("\nShapes:")
print("1D shape:", array_1d.shape)
print("2D shape:", array_2d.shape)
print("3D shape:", array_3d.shape)
```

```python
# Broadcasting example
marks = np.array([60, 70, 80, 90, 100])

bonus = 5

# NumPy automatically broadcasts 5 to every element
updated_marks = marks + bonus

print("Original marks:")
print(marks)

print("\nMarks after adding bonus:")
print(updated_marks)
```

```python
# Vectorised operations - no Python loops

scores = np.array([10, 20, 30, 40, 50])

print("Scores:", scores)
print("Scores × 2:", scores * 2)
print("Scores + 10:", scores + 10)
print("Scores squared:", scores ** 2)
print("Scores / 10:", scores / 10)
```

```python
# Matrix multiplication

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

result = A @ B

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

print("\nA @ B:")
print(result)
```

```python
# Create a student performance dataset

data = {
    "Student_ID": range(1, 21),
    "Age": [18, 19, 18, 20, 19, 18, 21, 20, 19, 18,
            20, 21, 19, 18, 20, 19, 21, 20, 18, 19],
    "Math_Score": [78, 85, 67, 90, 72, 88, 95, 64, 76, 81,
                   69, 92, 74, 86, 79, 68, 91, 83, 71, 87],
    "Python_Score": [82, 89, 70, 94, 75, 91, 97, 68, 79, 85,
                     73, 95, 78, 90, 83, 71, 94, 88, 76, 92]
}

df = pd.DataFrame(data)

# Save as CSV
df.to_csv("student_performance.csv", index=False)

# Display first 10 records
print(df.head(10))
```

```python
# Load the CSV dataset

df = pd.read_csv("student_performance.csv")

print("Dataset shape:", df.shape)

print("\nDataset:")
display(df)
```

```python
# Convert score columns to NumPy arrays

math_scores = df["Math_Score"].to_numpy()
python_scores = df["Python_Score"].to_numpy()

print("Math scores:")
print(math_scores)

print("\nPython scores:")
print(python_scores)
```

```python
# Convert score columns to NumPy arrays

math_scores = df["Math_Score"].to_numpy()
python_scores = df["Python_Score"].to_numpy()

print("Math scores:")
print(math_scores)

print("\nPython scores:")
print(python_scores)
```

```python
# Calculate standard deviation using NumPy

math_std = np.std(math_scores)
python_std = np.std(python_scores)

print("Math Score Standard Deviation:", math_std)
print("Python Score Standard Deviation:", python_std)
```

```python
# Calculate Pearson correlation using NumPy

correlation_matrix = np.corrcoef(math_scores, python_scores)

correlation = correlation_matrix[0, 1]

print("Correlation between Math and Python scores:", correlation)
```
