```python
# ============================================
# W1D3 SECTION 1: DATA LOADING
# ============================================

import numpy as np
import pandas as pd

print("NumPy version:", np.__version__)
print("Pandas version:", pd.__version__)

# Create an Indian student performance dataset
data = {
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": [
        "Aarav", "Priya", "Rahul", "Ananya", "Vikram",
        "Sneha", "Arjun", "Kavya", "Rohan", "Meera"
    ],
    "State": [
        "Karnataka", "Tamil Nadu", "Kerala", "Maharashtra",
        "Karnataka", "Kerala", "Tamil Nadu", "Karnataka",
        "Maharashtra", "Kerala"
    ],
    "Age": [20, 21, 20, 22, 21, 20, 23, 21, 22, 20],
    "Math_Score": [85, 90, 78, 88, 95, 82, 76, 91, 87, 89],
    "Python_Score": [88, 92, 80, 85, 94, 79, 74, 96, 83, 91]
}

# Create DataFrame
df = pd.DataFrame(data)

# Save as CSV
df.to_csv("indian_student_data.csv", index=False)

# Load CSV using Pandas
df = pd.read_csv("indian_student_data.csv")

print("\nDataset loaded successfully!")
print("Shape:", df.shape)

print("\nFirst 10 rows:")
display(df.head(10))
```

```python
# ============================================
# W1D3 SECTION 2: DATA INSPECTION
# ============================================

print("DATA TYPES")
print("=" * 40)
print(df.dtypes)

print("\nDATASET INFORMATION")
print("=" * 40)
df.info()

print("\nSTATISTICAL SUMMARY")
print("=" * 40)
display(df.describe())

print("\nMISSING VALUES")
print("=" * 40)
print(df.isnull().sum())

print("\nDUPLICATE ROWS")
print("=" * 40)
print(df.duplicated().sum())
```

```python
# ============================================
# W1D3 SECTION 3: DATA CLEANING
# ============================================

# Create a copy so the original data remains unchanged
cleaned_df = df.copy()

# Check missing values before cleaning
print("Missing values before cleaning:")
print(cleaned_df.isnull().sum())

# Remove duplicate rows
cleaned_df = cleaned_df.drop_duplicates()

# Remove rows containing missing values
cleaned_df = cleaned_df.dropna()

# Make sure numeric columns contain numeric values
cleaned_df["Age"] = pd.to_numeric(
    cleaned_df["Age"],
    errors="coerce"
)

cleaned_df["Math_Score"] = pd.to_numeric(
    cleaned_df["Math_Score"],
    errors="coerce"
)

cleaned_df["Python_Score"] = pd.to_numeric(
    cleaned_df["Python_Score"],
    errors="coerce"
)

# Remove any rows that became invalid
cleaned_df = cleaned_df.dropna()

print("\nData after cleaning:")
display(cleaned_df)

print("\nMissing values after cleaning:")
print(cleaned_df.isnull().sum())

print("\nDuplicate rows after cleaning:")
print(cleaned_df.duplicated().sum())
```

```python
# ============================================
# W1D3 SECTION 4: NUMPY OPERATIONS
# ============================================

# Convert Math scores to a NumPy array
math_scores = cleaned_df["Math_Score"].to_numpy()

print("NumPy Array:")
print(math_scores)

print("\nArray Shape:")
print(math_scores.shape)

print("\nArray Data Type:")
print(math_scores.dtype)

# SLICING
# Select the first five scores
first_five = math_scores[:5]

print("\nFirst five scores using slicing:")
print(first_five)

# BROADCASTING
# Add 5 bonus marks to every student's Math score
bonus_scores = math_scores + 5

print("\nScores after adding 5 bonus marks:")
print(bonus_scores)

# NumPy mathematical functions
print("\nMean Math Score:")
print(np.mean(math_scores))

print("\nMaximum Math Score:")
print(np.max(math_scores))

print("\nMinimum Math Score:")
print(np.min(math_scores))

print("\nStandard Deviation:")
print(np.std(math_scores))
```

```python
# ============================================
# W1D3 SECTION 5: TESTS AND FINAL VERIFICATION
# ============================================

# TEST 1: Check that DataFrame is not empty
assert not cleaned_df.empty, "DataFrame is empty!"

# TEST 2: Check required columns
required_columns = [
    "Student_ID",
    "Name",
    "State",
    "Age",
    "Math_Score",
    "Python_Score"
]

for column in required_columns:
    assert column in cleaned_df.columns, f"Missing column: {column}"

# TEST 3: Check for missing values
assert cleaned_df.isnull().sum().sum() == 0, \
    "Missing values still exist!"

# TEST 4: Check for duplicate rows
assert cleaned_df.duplicated().sum() == 0, \
    "Duplicate rows still exist!"

# TEST 5: Check NumPy array
assert isinstance(math_scores, np.ndarray), \
    "Math scores are not a NumPy array!"

print("All tests passed successfully!")

# Export cleaned dataset
cleaned_df.to_csv(
    "indian_student_data_cleaned.csv",
    index=False
)

print("\nCleaned dataset exported successfully!")

# Final summary
print("\n" + "=" * 50)
print("W1D3 - DATA LOADING, CLEANING & INSPECTION")
print("=" * 50)

print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", cleaned_df.shape)
print("Number of students:", len(cleaned_df))
print("Average Math Score:", round(np.mean(math_scores), 2))
print("Average Python Score:",
      round(cleaned_df["Python_Score"].mean(), 2))

print("\nGenerated files:")
print("1. indian_student_data.csv")
print("2. indian_student_data_cleaned.csv")

print("\nW1D3 COMPLETED SUCCESSFULLY!")
```
