```python
# ============================================
# W1D5 SECTION 1: SETUP AND DATA LOADING
# ============================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set Seaborn theme
sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
print("Seaborn version:", sns.__version__)

# Create an Indian student performance dataset
data = {
    "Student_ID": range(101, 121),

    "Name": [
        "Aarav", "Priya", "Rahul", "Ananya", "Vikram",
        "Sneha", "Arjun", "Kavya", "Rohan", "Meera",
        "Aditya", "Diya", "Kiran", "Neha", "Varun",
        "Ishita", "Manish", "Pooja", "Sanjay", "Riya"
    ],

    "State": [
        "Karnataka", "Tamil Nadu", "Kerala", "Maharashtra",
        "Karnataka", "Kerala", "Tamil Nadu", "Karnataka",
        "Maharashtra", "Kerala", "Karnataka", "Tamil Nadu",
        "Kerala", "Maharashtra", "Karnataka", "Kerala",
        "Tamil Nadu", "Karnataka", "Maharashtra", "Kerala"
    ],

    "Age": [
        20, 21, 20, 22, 21,
        20, 23, 21, 22, 20,
        21, 22, 20, 21, 23,
        20, 22, 21, 20, 22
    ],

    "Math_Score": [
        85, 90, 78, 88, 95,
        82, 76, 91, 87, 89,
        92, 84, 79, 86, 73,
        94, 81, 90, 88, 96
    ],

    "Python_Score": [
        88, 92, 80, 85, 94,
        79, 74, 96, 83, 91,
        95, 87, 82, 89, 70,
        97, 78, 93, 86, 98
    ]
}

# Create DataFrame
df = pd.DataFrame(data)

# Create Total and Average scores
df["Total_Score"] = df["Math_Score"] + df["Python_Score"]
df["Average_Score"] = df["Total_Score"] / 2

print("\nDataset loaded successfully!")
print("Shape:", df.shape)

display(df.head(10))
```

```python
# ============================================
# W1D5 SECTION 2: MATPLOTLIB DISTRIBUTIONS
# ============================================

# --------------------------------------------
# Histogram: Math Scores
# --------------------------------------------

plt.figure(figsize=(8, 5))

plt.hist(
    df["Math_Score"],
    bins=8,
    edgecolor="black"
)

plt.title("Distribution of Mathematics Scores")
plt.xlabel("Math Score")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()


# --------------------------------------------
# Histogram: Python Scores
# --------------------------------------------

plt.figure(figsize=(8, 5))

plt.hist(
    df["Python_Score"],
    bins=8,
    edgecolor="black"
)

plt.title("Distribution of Python Scores")
plt.xlabel("Python Score")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()


# --------------------------------------------
# Box Plot: Compare Scores
# --------------------------------------------

plt.figure(figsize=(8, 5))

plt.boxplot(
    [
        df["Math_Score"],
        df["Python_Score"]
    ],
    tick_labels=["Math", "Python"]
)

plt.title("Math vs Python Score Distribution")
plt.ylabel("Score")

plt.tight_layout()
plt.show()
```

```python
# ============================================
# W1D5 SECTION 4: CORRELATION HEATMAP
# ============================================

# Select numerical columns
numeric_columns = [
    "Age",
    "Math_Score",
    "Python_Score",
    "Total_Score",
    "Average_Score"
]

# Calculate correlation
correlation = df[numeric_columns].corr()

print("Correlation Matrix:")
display(correlation)


# Create heatmap
plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Student Performance")

plt.tight_layout()
plt.show()
```
