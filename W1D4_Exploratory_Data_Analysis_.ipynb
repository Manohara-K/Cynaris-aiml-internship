```python
# ============================================
# W1D4 SECTION 1: LOAD AND INSPECT DATA
# ============================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create an Indian student performance dataset
data = {
    "Student_ID": range(101, 121),
    "Name": [
        "Aarav", "Priya", "Rahul", "Ananya", "Vikram",
        "Sneha", "Arjun", "Kavya", "Rohan", "Meera",
        "Aditya", "Diya", "Kiran", "Neha", "Varun",
        "Ishita", "Manish", "Pooja", "Sanjay", "Riya"
    ],
    "State": [
        "Karnataka", "Tamil Nadu", "Kerala", "Maharashtra",
        "Karnataka", "Kerala", "Tamil Nadu", "Karnataka",
        "Maharashtra", "Kerala", "Karnataka", "Tamil Nadu",
        "Kerala", "Maharashtra", "Karnataka", "Kerala",
        "Tamil Nadu", "Karnataka", "Maharashtra", "Kerala"
    ],
    "Age": [
        20, 21, 20, 22, 21,
        20, 23, 21, 22, 20,
        21, 22, 20, 21, 23,
        20, 22, 21, 20, 22
    ],
    "Math_Score": [
        85, 90, 78, 88, 95,
        82, 76, 91, 87, 89,
        92, 84, 79, 86, 73,
        94, 81, 90, 88, 96
    ],
    "Python_Score": [
        88, 92, 80, 85, 94,
        79, 74, 96, 83, 91,
        95, 87, 82, 89, 70,
        97, 78, 93, 86, 98
    ]
}

# Create DataFrame
df = pd.DataFrame(data)

# Save as CSV
df.to_csv("indian_student_performance.csv", index=False)

# Load CSV
df = pd.read_csv("indian_student_performance.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nFirst 10 rows:")
display(df.head(10))
```

```python
# ============================================
# W1D4 SECTION 2: DATA INSPECTION
# ============================================

print("========== DATA INFORMATION ==========")
df.info()

print("\n========== STATISTICAL DESCRIPTION ==========")
display(df.describe())

print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())

print("\n========== DUPLICATE ROWS ==========")
print(df.duplicated().sum())
```

```python
# ============================================
# W1D4 SECTION 3B: CORRELATION HEATMAP
# ============================================

# Select numeric columns
numeric_df = df.select_dtypes(include=np.number)

# Calculate correlation matrix
correlation_matrix = numeric_df.corr()

print("Correlation Matrix:")
display(correlation_matrix)

# Plot heatmap
plt.figure(figsize=(8, 6))

plt.imshow(
    correlation_matrix,
    interpolation="nearest"
)

plt.colorbar()

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.title("Correlation Heatmap")

# Add correlation values to the heatmap
for i in range(len(correlation_matrix)):
    for j in range(len(correlation_matrix.columns)):
        plt.text(
            j,
            i,
            f"{correlation_matrix.iloc[i, j]:.2f}",
            ha="center",
            va="center"
        )

plt.tight_layout()
plt.show()
```

```python
# ============================================
# W1D4 SECTION 3C: TOP-10 CATEGORY COUNTS
# ============================================

# Count occurrences of each state
top_states = df["State"].value_counts().head(10)

print("Top 10 State Counts:")
display(top_states)

# Plot the top 10 categories
plt.figure(figsize=(9, 5))

top_states.plot(kind="bar")

plt.title("Top 10 State Counts")
plt.xlabel("State")
plt.ylabel("Number of Students")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# ============================================
# W1D4 SECTION 4: CLEANING AND TRANSFORMATION
# ============================================

# Create a copy for EDA processing
eda_df = df.copy()

# Remove duplicate records
eda_df = eda_df.drop_duplicates()

# Handle missing values
eda_df = eda_df.dropna()

# Create total score
eda_df["Total_Score"] = (
    eda_df["Math_Score"] +
    eda_df["Python_Score"]
)

# Create average score
eda_df["Average_Score"] = (
    eda_df["Math_Score"] +
    eda_df["Python_Score"]
) / 2

# Create performance category
eda_df["Performance"] = pd.cut(
    eda_df["Average_Score"],
    bins=[0, 60, 75, 90, 100],
    labels=["Needs Improvement", "Average", "Good", "Excellent"],
    include_lowest=True
)

print("Transformed DataFrame:")
display(eda_df.head(10))

print("\nPerformance Distribution:")
display(eda_df["Performance"].value_counts())
```

```python
# ============================================
# W1D4 SECTION 5: TESTS AND FINAL CHECK
# ============================================

# Test 1: DataFrame should not be empty
assert not eda_df.empty, "EDA DataFrame is empty!"

# Test 2: Required columns must exist
required_columns = [
    "Student_ID",
    "Name",
    "State",
    "Age",
    "Math_Score",
    "Python_Score",
    "Total_Score",
    "Average_Score",
    "Performance"
]

for column in required_columns:
    assert column in eda_df.columns, f"Missing column: {column}"

# Test 3: No missing values
assert eda_df.isnull().sum().sum() == 0, \
    "Missing values found!"

# Test 4: No duplicates
assert eda_df.duplicated().sum() == 0, \
    "Duplicate rows found!"

# Test 5: Score values should be between 0 and 100
assert eda_df["Math_Score"].between(0, 100).all(), \
    "Invalid Math score found!"

assert eda_df["Python_Score"].between(0, 100).all(), \
    "Invalid Python score found!"

print("All tests passed successfully!")

print("\n" + "=" * 55)
print("W1D4 EXPLORATORY DATA ANALYSIS COMPLETED")
print("=" * 55)

print("\nOriginal shape:", df.shape)
print("Cleaned shape:", eda_df.shape)

print("\nAverage Math Score:",
      round(eda_df["Math_Score"].mean(), 2))

print("Average Python Score:",
      round(eda_df["Python_Score"].mean(), 2))

print("\nMissing values:")
print(eda_df.isnull().sum())

print("\nDuplicates:", eda_df.duplicated().sum())
```
